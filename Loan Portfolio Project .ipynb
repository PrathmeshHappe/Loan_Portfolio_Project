```python
df.rename(columns= {
    "defaults_on_file":"Defaults_on_File",
    "debt_to_income_ratio":"debt_ratio",
    "loan_to_income_ratio":"loan_income"
},inplace = True)
```

```python
df.drop(columns = ["Defaults_on_File"]),
df.drop(columns = ["delinquencies_last_2yrs"]),
df.drop(columns = ["derogatory_marks"]),
df.drop(columns = ["debt_ratio"]),
df.drop(columns = ["loan_income"]),
df.drop(columns = ["payment_to_income_ratio"]).head()
```

```python
df["loan_intent"] = df["loan_intent"].str.upper()
```

```python
loan_map= {
    0 : "Rejected",
    1 : "Approved"
}
```

```python
df["approved/rejected"] = df["loan_status"].map(loan_map)
```

```python
df.sample(5)
```

```python
pd.pivot_table(df, index = "age", columns = "occupation_status", values = "credit_score", aggfunc = "mean").head(10)
```

```python
sns.countplot(data=df, x='approved/rejected')
plt.title('Loan Status Distribution')
plt.show()
```

```python
sns.countplot(data=df, x='occupation_status', hue='approved/rejected')
plt.title('Loan Status by Occupation')
plt.show()
```

```python
sns.countplot(data=df, x='product_type', hue='approved/rejected')
plt.title('Loan Status by Product Type')
plt.xticks(rotation=0)
plt.show()
```

```python
sns.countplot(data=df, x='loan_intent')
plt.title('Loan Intent Distribution')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.histplot(df['annual_income'], bins=10, kde=True)
plt.title('Annual Income Distribution')
plt.show()
```

```python
sns.boxplot(data=df, x='loan_status', y='annual_income')
plt.title('Annual Income by Loan Status')
plt.show()
```

```python
sns.scatterplot(data=df, x='annual_income', y='loan_amount', hue='approved/rejected')
plt.title('Annual Income vs Loan Amount')
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Loan Dataset')
plt.show()
```

```python
df['product_type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Loan Product Distribution')
plt.ylabel('')
plt.show()
```
